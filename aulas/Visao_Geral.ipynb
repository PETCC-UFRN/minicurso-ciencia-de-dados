## Introdução a Ciência de Dados



Você provavelmente já ouviu falar que as Ciências de Dados são a intersecção entre a Computação e a Estatística, mas, na verdade, a área abrange muito mais do que isto.

No contexto atual, a área de Ciência de Dados se encontra como o meio termo entre Ciência da Computação, Matemática e Estatística e Especialização Científica/Conhecimento de Negócio.

Além disso, atualmente estamos na chamada como "Era do *Big Data*", e, assim, a demanda por profissionais que saibam manipular essa enorme quantidade de dados se torna cada vez maior.

ciencia de dados venn.png

#### Atribuições de um Cientista de Dados:

- Conhecer o suficiente sobre os negócios para fazer perguntas pertinentes e identificar pontos problemáticos.

- Aplicar estatísticas e ciência da computação, juntamente com conhecimento sobre negócios, à análise de dados.

- Usar uma grande variedade de ferramentas e técnicas para preparar e extrair dados, desde bancos de dados e SQL até métodos de mineração de dados e de integração de dados.

- Extrair insights de big data usando análise preditiva e inteligência artificial (IA), incluindo modelos de machine learning, processamento de linguagem natural e deep learning.

- Contar e ilustrar histórias que transmitam claramente os resultados para os tomadores de decisões e stakeholders em todos os níveis de conhecimento técnico.

- Colaborar com outros membros da equipe de ciência de dados, como:
  - analistas de dados e negócios,
  - arquitetos de TI,
  - engenheiros de dados e
  - desenvolvedores de aplicativos.

#### Etapas de um projeto de Ciência de Dados

- **Limpeza**: Pré-processamento dos dados
- **Análise exploratória**: Extrair informações dos dados a partir de gráficos e estatísticas e levantar hipóteses
- **Datamining**: Extrair informações dos dados utilizando modelos de Aprendizado de Máquina e levantar hipóteses
- **Hipóteses e Estatística**: Provar ou refutar as hipóteses levantadas ao longo do projeto com o cálculo estatístico.

# Etapas de um projeto de Ciência de Dados

Um projeto de Ciência de Dados é dividido em diversas etapas, cada uma cumprindo sua função no objetivo geral: extrair informações sobre uma base de dados.

Mas antes de falar sobre essas etapas na prática, vamos falar sobre pontos importantes em qualquer prática envolvendo Ciência de Dados!

---

### 1. Conhecer melhor o contexto dos seus dados

Uma parte importantíssima da análise de uma base de dados é entender bem o contexto no qual aqueles dados se aplicam, para ter uma melhor noção de que decisões tomar em cada etapa da análise.

Ferramentas de pesquisa são suas melhores amigas nesse momento, e quanto mais tempo passar coletando conhecimento sobre a área a ser estudada, melhor!

### 2. Senso crítico

Sendo um adendo ao primeiro item, sempre questionar as informações que aparecem no decorrer da análise é muito importante - incluindo até, muitas vezes, questionar os próprios dados! às vezes são encontradas incorretudes, e, como cientista de dados, é sua função corrigi-las!

### 3. Imparcialidade

É muito tentador manter nossos viéses quando estamos no meio da análise de uma base de dados, e uma das virtudes de um bom cientista de dados é saber ser imparcial e ficar satisfeito com as conclusões que os próprios dados dão. Podemos criar hipóteses na hora de explorar melhor os dados, para termos metas mais bem definidas, mas as vezes passamos mais tempo tentando "provar" algo que realmente extraindo informações úteis da base de dados!

### 4. Perseverança

A análise de dados pode parecer simples no papel, mas é fácil acabar estagnando na prática, principalmente no começo, por não conhecer muito bem as ferramentas e não ter muita experiência manipulando dados. Às vezes, uma decisão feita em uma etapa inicial pode impactar (e até inviabilizar!) uma etapa mais ao final da análise - o que pode resultar na necessidade de voltar à etapa onde o erro foi cometido, e refazer tudo do zero...
