## O Desafio do Dia 1: Limpeza de Dados

No primeiro dia de qualquer projeto de Ciência de Dados, a limpeza e pré-processamento dos dados são etapas cruciais. Dados do mundo real raramente vêm perfeitos; eles podem conter valores ausentes, inconsistências, formatos incorretos e outliers que podem comprometer a qualidade das análises subsequentes. O objetivo deste primeiro desafio é garantir que o nosso conjunto de dados de turmas da UFRN esteja pronto para as próximas etapas de análise.

Vocês deverão baixar os arquivo CSV contendo os dados brutos das turmas da UFRN dos anos de 2023 e 2024 no site oficial:https://dados.ufrn.br/dataset/turmas . O dicionário de dados, que descreve cada coluna e seu significado, também é fornecido e deve ser a sua principal referência para entender a estrutura dos dados.

## Objetivos de Aprendizagem

Ao final deste dia, vocês deverão ser capazes de:

* **Carregar e Inspecionar Dados:** Importar o conjunto de dados para um ambiente Python (preferencialmente Jupyter Notebook) e realizar uma inspeção inicial para entender sua estrutura, tipos de dados e a presença de valores ausentes.
* **Identificar e Tratar Valores Ausentes:** Localizar colunas com valores ausentes e aplicar estratégias adequadas para tratá-los (ex: remoção de linhas/colunas, preenchimento com média/mediana/moda, ou imputação mais avançada), justificando suas escolhas.
* **Corrigir Tipos de Dados Incorretos:** Assegurar que cada coluna tenha o tipo de dado apropriado (numérico, categórico, data, booleano), realizando conversões quando necessário.
* **Lidar com Dados Duplicados:** Identificar e remover registros duplicados que possam existir no dataset.
* **Padronizar e Normalizar Dados (se aplicável):** Para colunas textuais, verificar a necessidade de padronização (ex: converter para minúsculas, remover caracteres especiais). Para colunas numéricas, considerar a normalização ou padronização se houver grande variação de escala e se for relevante para futuras análises.
* **Identificar e Tratar Outliers (inicialmente):** Realizar uma análise exploratória inicial para identificar possíveis outliers em colunas numéricas e propor estratégias para lidar com eles, se necessário.
* **Documentar o Processo:** Registrar todas as etapas de limpeza e as decisões tomadas no notebook, explicando o porquê de cada ação.

## Tarefas a Serem Realizadas no Notebook

Crie um novo Jupyter Notebook e siga os passos abaixo:

1.  **Carregamento dos Dados:** Carregue o arquivo de dados das turmas da UFRN para um DataFrame Pandas.
2.  **Visão Geral dos Dados:** Utilize funções como `.info()`, `.describe()`, `.head()`, `.isnull().sum()` para obter um panorama inicial do dataset.
3.  **Tratamento de Valores Ausentes:** Implemente as estratégias de tratamento de valores ausentes que você julgar mais adequadas para cada coluna, baseando-se no dicionário de dados e na natureza da informação.
4.  **Ajuste de Tipos de Dados:** Converta os tipos de dados das colunas que estiverem incorretos (ex: datas como strings, números como objetos).
5.  **Remoção de Duplicados:** Verifique e remova quaisquer linhas duplicadas que possam existir.
6.  **Padronização de Texto:** Para colunas textuais como `nivel_ensino` ou `campus_turma`, aplique padronizações básicas (ex: `.str.lower().str.strip()`).
7.  **Análise e Tratamento de Outliers:** Faça uma análise visual (ex: boxplots) ou estatística para identificar outliers e decida como tratá-los (remoção, transformação, etc.), se for o caso.
8.  **Verificação Final:** Após todas as etapas, inspecione novamente o dataset para garantir que a limpeza foi bem-sucedida e que os dados estão prontos para a próxima fase.
