# Data Mining e Clustering, extraindo novas informações.

## Mineração de Dados

A mineração de dados, ou data mining, é o processo de explorar grandes conjuntos de dados para encontrar padrões, correlações e tendências ocultas. O seu principal objetivo é transformar dados brutos em conhecimento útil e acionável. Ao fazer isso, ela permite que empresas e organizações tomem decisões mais inteligentes e estratégicas, baseadas em evidências concretas encontradas nos próprios dados que coletam.

É útil pensar na ciência de dados como um campo amplo. Dentro dele, a mineração de dados é a etapa específica de descoberta de padrões. Para realizar essa tarefa, ela frequentemente utiliza ferramentas do aprendizado de máquina (machine learning). Em resumo, o aprendizado de máquina fornece os algoritmos que automatizam e potencializam a busca por insights valiosos durante o processo de mineração de dados.

## O Processo de Mineração de Dados: Tarefas e Aplicações

Agora que entendemos o que é a mineração de dados e sua relação com a ciência de dados, vamos mergulhar em suas aplicações práticas e nas tarefas mais comuns que um cientista de dados realiza.

### Aplicações Práticas



A mineração de dados é uma ferramenta versátil, aplicada em inúmeros setores para resolver problemas complexos:

* Negócios e Finanças: Análise de risco de crédito, detecção de transações fraudulentas, previsão de vendas, otimização de campanhas de marketing e análise de sentimento de clientes sobre uma marca.
* Saúde: Diagnóstico precoce de doenças com base em exames e histórico do paciente, identificação de fatores de risco para epidemias, otimização da gestão de leitos em hospitais e descoberta de padrões em testes de novos medicamentos.
* E-commerce e Varejo: Sistemas de recomendação ("Quem comprou A também comprou B"), análise de cesta de compras para otimização de layout de loja, segmentação de clientes para promoções personalizadas e gestão de estoque.



### Tarefas Comuns na mineração de dados.



As aplicações que vimos são, na prática, executadas por meio de algumas tarefas fundamentais. As mais importantes são:

* **Classificação**: O objetivo é prever uma categoria ou classe. O algoritmo aprende com dados já rotulados para classificar novos dados.
  * Exemplo: Determinar se um e-mail é spam ou não spam.
* **Regressão**: Similar à classificação, mas o objetivo é prever um valor numérico contínuo.
  * Exemplo: Prever o preço de um imóvel com base em sua área, localização e número de quartos.
* **Clusterização (Agrupamento)**: Agrupa dados semelhantes em "clusters" ou segmentos, sem conhecimento prévio das categorias. O objetivo é encontrar a estrutura inerente nos dados.
  * Exemplo: Agrupar clientes de uma loja em segmentos (ex: compradores frequentes, compradores de alto valor, visitantes ocasionais).
* **Regras de Associação**: Descobre relações interessantes entre variáveis em grandes bancos de dados.
  * Exemplo: A famosa análise que descobriu que homens que compravam fraldas às sextas-feiras também tendiam a comprar cerveja.
* **Detecção de Anomalias**: Identifica itens, eventos ou observações que não se conformam a um padrão esperado.
  * Exemplo: Detectar uma compra com cartão de crédito que foge completamente do padrão de gastos de um cliente.

### Aprendizado Supervisionado vs. Não Supervisionado

Todas essas tarefas se encaixam em duas categorias principais de aprendizado de máquina:

* **Aprendizado Supervisionado**: Ocorre quando treinamos nosso modelo com um conjunto de dados que já contém as "respostas corretas" (rótulos ou etiquetas). O algoritmo aprende a mapear a entrada para a saída.

  * Tarefas: Classificação e Regressão.
* **Aprendizado Não Supervisionado**: Aqui, o algoritmo trabalha com dados que não possuem rótulos. O objetivo é explorar os dados e encontrar padrões ou estruturas por conta própria.

  * Tarefas: Clusterização e Regras de Associação.
A Detecção de Anomalias pode ser tanto supervisionada quanto não supervisionada, dependendo da abordagem.

### Regressão

Na regressão, buscamos prever um valor numérico. No exemplo abaixo, usamos o comprimento da pétala para prever sua largura. O gráfico mostra os dados reais (pontos) e uma linha de regressão que representa o modelo. Essa linha é a melhor aproximação para estimar a largura da pétala com base em seu comprimento.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.linear_model import LinearRegression

# Carregar dados (pode ser omitido se a célula anterior já foi executada)
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Usaremos o comprimento e a largura da pétala para ilustrar a regressão
X = iris_df[['petal length (cm)']]
y = iris_df['petal width (cm)']

# Treinar um modelo de regressão linear simples
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)

# Gerar o gráfico de regressão
plt.figure(figsize=(8, 6))
sns.scatterplot(x='petal length (cm)', y='petal width (cm)', data=iris_df, label='Dados Reais', s=100, alpha=0.8)
plt.plot(iris_df['petal length (cm)'], y_pred, color='red', linewidth=2, label='Linha de Regressão')
plt.title('Exemplo de Regressão: Largura vs. Comprimento da Pétala')
plt.xlabel('Comprimento da Pétala (cm)')
plt.ylabel('Largura da Pétala (cm)')
plt.legend()
plt.grid(True)
plt.show()
```

### Clusterização

Na clusterização, o algoritmo agrupa os dados sem conhecer as espécies. Usamos o algoritmo K-Means para encontrar 3 grupos (clusters) com base na semelhança entre as flores (comprimento e largura da pétala). A cor de cada ponto representa o cluster atribuído pelo algoritmo. Note como os grupos encontrados pelo K-Means são muito semelhantes às espécies reais!



```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

# Carregar dados (pode ser omitido se as células anteriores já foram executadas)
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Usaremos os mesmos dados, mas sem os rótulos das espécies
features = iris_df[['petal length (cm)', 'petal width (cm)']]

# Aplicar K-Means com 3 clusters
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
clusters = kmeans.fit_predict(features)
iris_df['cluster'] = clusters

# Gerar o gráfico de clusterização
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x='petal length (cm)',
    y='petal width (cm)',
    hue='cluster',
    data=iris_df,
    palette='viridis',
    s=100,
    alpha=0.8,
    legend='full'
)
# Plotar os centroides
centroids = kmeans.cluster_centers_
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', s=250, marker='X', label='Centroides')
plt.title('Exemplo de Clusterização (K-Means) em Dados de Pétalas')
plt.xlabel('Comprimento da Pétala (cm)')
plt.ylabel('Largura da Pétala (cm)')
plt.legend()
plt.grid(True)
plt.show()
```

## Clusterização

No contexto de ciência de dados, clusterização é uma tarefa de aprendizado não supervisionado que consiste em **agrupar um conjunto de objetos** de modo que objetos no mesmo grupo (chamado de *cluster*) sejam mais **similares entre si** do que com aqueles em outros grupos.


O objetivo é **descobrir estruturas ou padrões inerentes nos dados** sem qualquer conhecimento prévio sobre as categorias existentes, uma vez que o algoritmo não recebe exemplos rotulados de como os grupos devem ser.


Assim como na classificação, cada instância é atribuída a um grupo. Contudo, na tarefa de aprendizado supervisionado, os grupos já estão pré-definidos e o objetivo da classificação é a predição de classes; enquanto a clusterização visa a descoberta de estruturas inerentes nos dados com base na semelhança entre as instâncias.

## Exemplos de aplicação da clusterização

A clusterização é utilizada em um leque amplo de áreas de atuação, incluindo:

* **Segmentação de clientes**: É possível agrupar clientes com base nas compras e atividades em um site. Isso ajuda a compreender o perfil dos usuários e suas respectivas necessidades, a fim de adaptar produtos e campanhas de marketing a cada segmento.

* **Detecção de anomalias**: Qualquer instância que tenha uma baixa afinidade com todos os *clusters* provavelmente será um outlier. Utilizando o exemplo anterior, se a clusterização foi feita com base no comportamento de usuários em um site, é possível detectar clientes com comportamentos discrepantes, como um número incomum de solicitações por segundo.

* **Segmentação de imagens**: Ao clusterizar pixels de acordo com a cor e substituir a cor de cada pixel pela cor média do cluster, é possível reduzir significativamente o número de cores diferentes em uma imagem.

## Noções de similaridade e dissimilaridade

Antes de apresentar métricas utilizadas pelos algoritmos de clusterização, precisamos pensar em como nós, humanos, decidimos que duas coisas são parecidas (similares) ou diferentes (dissimilares).

Imagine que você está caminhando em um parque e acaba tropeçando em uma planta que nunca viu antes. Olhando ao redor, você nota mais algumas plantas e, apesar de não serem idênticas à primeira planta, são suficientemente parecidas para você saber que provavelmente pertencem à mesma espécie (ou ao mesmo gênero, pelo menos). Não é necessário um botânico para identificar que estas plantas são da mesma espécie (são similares) e não são árvores (a planta é diferente de uma árvore), por exemplo. Estas noções de similaridade e dissimilaridade são fundamentais para a clusterização.

Contudo, não podemos depender da intuição em ciência de dados. É preciso encontrar uma maneira de quantificar essas semelhanças e diferenças entre instâncias. É aqui que entram as métricas: elas indicam o quão longe dois objetos estão um do outro em um espaço de dados.

A escolha da métrica mais adequada depende do que está sendo medido e do que a "distância" significa no contexto fornecido. As principais métricas são:

* **Distância Euclidiana**: Mede a distância em linha reta entre dois pontos em um espaço n-dimensional.

* **Distância de Manhattan**: Mede a distância somando os comprimentos dos caminhos percorridos em cada eixo.

* **Similaridade de Cosseno**: Mede o ângulo entre dois vetores para determinar se apontam na mesma direção, ignorando tamanho ou magnitude.

Na célula abaixo, temos uma visualização dessas 3 métricas:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import distance

# dois pontos A e B
A = np.array([3, 4])
B = np.array([7, 1])

# cálculo das distâncias
dist_euclid = distance.euclidean(A, B)
dist_manhattan = distance.cityblock(A, B)
sim_cosseno = 1 - distance.cosine(A, B)

# visualização
fig, axs = plt.subplots(1, 3, figsize=(15, 5))

# distância euclidiana
axs[0].scatter(*A, color='blue', label='Ponto A (3,4)', s=100)
axs[0].scatter(*B, color='red', label='Ponto B (7,1)', s=100)
axs[0].plot([A[0], B[0]], [A[1], B[1]], 'k--', label=f'Distância Euclidiana = {dist_euclid}')
axs[0].set_title('Distância Euclidiana')
axs[0].legend()
axs[0].grid(True)
axs[0].set_xlim(2, 8)
axs[0].set_ylim(0, 6)

# distância de manhattan
axs[1].scatter(*A, color='blue', label='Ponto A (3,4)', s=100)
axs[1].scatter(*B, color='red', label='Ponto B (7,1)', s=100)
# caminho
axs[1].plot([A[0], B[0]], [A[1], A[1]], 'g--')
axs[1].plot([B[0], B[0]], [A[1], B[1]], 'g--', label=f'Distância de Manhattan = {dist_manhattan}')
axs[1].set_title('Distância de Manhattan')
axs[1].legend()
axs[1].grid(True)
axs[1].set_xlim(2, 8)
axs[1].set_ylim(0, 6)

# similaridade de cosseno
origin = [0, 0]
axs[2].quiver(*origin, *A, angles='xy', scale_units='xy', scale=1, color='blue', label='Ponto A')
axs[2].quiver(*origin, *B, angles='xy', scale_units='xy', scale=1, color='red', label='Ponto B')
axs[2].set_xlim(0, 8)
axs[2].set_ylim(0, 6)
axs[2].set_title(f'Similaridade de Cosseno = {sim_cosseno:.2f}')
axs[2].legend()
axs[2].grid(True)

plt.suptitle('Visualização das Métricas de Distância e Similaridade', fontsize=16)
plt.tight_layout()
plt.show()
```

## Algoritmos de Clusterização

Apesar de introduzirmos alguns conceitos sobre a clusterização, não existe uma
definição universal do que é um *cluster* : depende do contexto, do objetivo da análise e dos diferentes algoritmos identificarem tipos distintos de *cluster*. Isso acontece porque diferentes algoritmos de clusterização são baseados em diferentes noções do que forma um *cluster*. De maneira geral, os algoritmos se dividem nos tipos:

* Particionais: Divide os dados em um número pré-determinado de grupos (*k*). Os algoritmos mais famosos são:
  * K-means: agrupa os dados tentando encontrar os centroides de cada grupo.
  * K-medoids: semelhante ao k-means, escolhe sempre uma instância real do próprio grupo para ser o centro (medoide).

* Hierárquicos: Organiza os dados em uma estrutura semelhante a uma árvore, chamada de dendograma. São separados em:
  * Aglomerativo: cada ponto de dado é seu próprio cluster e vai fundindo os mais próximos.
  * Divisivo: todos os dados começam em um único cluster e são divididos recursivamente.

* Baseados em densidade: Define *clusters* como áreas de alta densidade de pontos, separadas por áreas de baixa densidade. O principal algoritmo é o DBSCAN (Density-Based Spatial Clustering of Applications with Noise).

### K-Means e DBSCAN

O K-Means é frequentemente o primeiro algoritmo de clusterização a ser aprendido, pois é simples e intuitivo. Ele assume que um "bom" cluster é um grupo de pontos que estão próximos de um ponto central, o centroide. Assim, o algoritmo tenta minimizar a soma das distâncias de todos os pontos de dados até o centroide do seu *cluster*.

É um algoritmo relativamente simples de entender e implementar, funciona bem com um grande número de amostras e apresenta resultados fáceis de interpretar. No entanto, apresenta alguns pontos fracos, como:

* A pré-definição de um número *k* de *clusters*.

* Sensibilidade aos centroides iniciais.

* Não lida bem com *clusters* de formatos irregulares, aninhados ou alongados.

* Não lida bem com outliers.


Com isso, o DBSCAN (Density-Based Spatial Clustering of Applications with Noise) aparece como uma alternativa que atua nas principais fraquezas do K-Means. Pela sua natureza baseada em densidade, o DBSCAN assume que um cluster é uma área densa de pontos, descartando a ideia de centroide utilizada pelo K-Means. Com isso, o algoritmo conecta pontos que estão próximos uns dos outros e separa as regiões de alta densidade das regiões de baixa densidade.

Estes dois algoritmos mencionados são os principais algoritmos de clusterização. Focaremos no K-Means pois ele oferece simplicidade teórica, utilidade prática e certa profundidade conceitual, atuando como ponte para tópicos mais avançados de ciência de dados.

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# criação de dados para simulação
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

# visualização dos dados
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Dados simulados para exemplificar clusterização")
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans

# utilizando k-means com k=4
kmeans = KMeans(n_clusters=4, random_state=42)
y_kmeans = kmeans.fit_predict(X)

# plotando resultado
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', s=200, alpha=0.75, marker='X', label='Centroides')
plt.title("Clusterização K-Means (k=4)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import DBSCAN

# DBSCAN
# eps: raio
# min_samples: densidade mínima
dbscan = DBSCAN(eps=0.8, min_samples=5)
y_dbscan = dbscan.fit_predict(X)

# plotando resultados
plt.scatter(X[:, 0], X[:, 1], c=y_dbscan, cmap='Paired', s=50)
plt.title("Clusterização DBSCAN")
plt.grid(True)
plt.show()
```

## K-Means

Como enunciado previamente, o K-Means é o algoritmo de clusterização particional mais famoso. O seu objetivo é particionar os dados em *k* grupos distintos, onde cada grupo é representado por um ponto central chamado centroide.

### Visão Geral
O primeiro passo do algoritmo é inicializar aleatoriamente os centroides. Ele seleciona aleatoriamente *k* pontos de dados do próprio conjunto de dados e esses *k* pontos serão designados como os centroides iniciais. Depois da inicialização, o algoritmo entra em um ciclo de dois passos que se repete até a convergência:

* Atribuição: O algoritmo calcula a distância de cada ponto de dado a cada um dos *k* centroides e este ponto de dado é atribuído ao cluster do centroide ao qual ele tem a menor distância

* Atualização: Depois de todos os pontos serem atribuídos a um cluster, os centroides passam por uma reavaliação. Para cada cluster, o algoritmo calcula a média de todos os pontos que foram atribuídos a ele e o resultado dessa média se torna a nova localização do centroide para aquele cluster.

O algoritmo repete os passos de atribuição e atualização e o ciclo se encerra quando uma das condições é atingida:

* Os centroides param de se mover significativamente.
* As atribuições dos pontos de dados aos *clusters* não mudam mais entre uma iteração e outra.
* Um número máximo de iterações é atingido.

Quando o ciclo para, dizemos que o algoritmo **convergiu** e os *clusters* estão definidos.


### Como determinar o número K?

O K-Means exige que o usuário defina o valor de *k* antes de executá-lo. Se você escolher um número *k* muito baixo, o algoritmo pode agrupar *clusters* que na verdade são distintos. Se você escolher um número *k* muito alto, pode dividir *clusters* que são naturalmente um só. Assim, como descobrir o número ideal de agrupamentos? Não existe uma resposta única, mas existem métodos para embasar a sua tomada de decisão. Os mais famosos são:

* Método do Cotovelo: se baseia na ideia de inércia. A inércia mede o quão compactos os *clusters* são — quanto menor a inércia, mais denso e compacto é o *cluster*.

* Análise de Silhueta: mede a semelhança de um ponto de dados dentro de seu cluster (coesão) em comparação com outros *clusters* (separação). A semelhança é representada por um "score" de silhueta para cada ponto de dado, que varia de [-1, +1].


Vamos retornar ao dataset Iris para aplicar os métodos descritos acima.

```python
from sklearn.preprocessing import StandardScaler

# carregando dataset Iris
iris = load_iris()
X = iris.data
y_true = iris.target
feature_names = iris.feature_names
target_names = iris.target_names

# padronização dos dados
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# método do cotovelo

# valores de k
k_range = range(1, 11)
inertias = []

for k in k_range:
  kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
  kmeans.fit(X_scaled)
  inertias.append(kmeans.inertia_)

plt.figure(figsize=(8,4))
plt.plot(k_range, inertias, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Número de Clusters (k)', fontsize=12)
plt.ylabel('Inércia', fontsize=12)
plt.title('Método do Cotovelo para Determinar k Ótimo', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)

plt.axvline(x=3, color='red', linestyle='--', alpha=0.7, label='k=3 (Real)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

# método da silhueta
k_range_silhouette = range(2, 11)
silhouette_scores = []

for k in k_range_silhouette:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)
    print(f"k={k}: Silhouette Score = {score:.3f}")

best_k = k_range_silhouette[np.argmax(silhouette_scores)]
best_score = max(silhouette_scores)

print(f"\nMelhor k segundo Silhueta: {best_k} (Score: {best_score:.3f})")

plt.figure(figsize=(8, 4))
plt.plot(k_range_silhouette, silhouette_scores, 'go-', linewidth=2, markersize=8)
plt.xlabel('Número de Clusters (k)', fontsize=12)
plt.ylabel('Silhouette Score', fontsize=12)
plt.title('Análise de Silhueta para Determinar k Ótimo', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)

plt.axvline(x=best_k, color='red', linestyle='--', alpha=0.7, label=f'Melhor k={best_k} (Score={best_score:.3f})')
plt.legend()
plt.tight_layout()
plt.show()
```
