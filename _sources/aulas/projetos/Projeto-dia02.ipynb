## O Desafio do Dia 2: Explorando os Dados das Turmas da UFRN

Hoje, vocês irão mergulhar profundamente nos dados limpos que prepararam ontem. O objetivo é compreender a distribuição das variáveis, identificar padrões interessantes, explorar relações entre diferentes características das turmas e formular hipóteses que poderão ser testadas nos próximos dias.

A análise exploratória não é apenas sobre criar gráficos bonitos, é sobre fazer as perguntas certas aos dados e buscar respostas que possam gerar valor e insights acionáveis para a gestão acadêmica da UFRN.

### Objetivos de Aprendizagem

Ao final deste dia, vocês deverão ser capazes de:

1. **Realizar Análise Univariada**: Examinar cada variável individualmente, compreendendo sua distribuição, valores centrais, dispersão e identificando características peculiares.  
2. **Conduzir Análise Bivariada**: Explorar relações entre pares de variáveis, identificando correlações, associações e padrões de dependência.  
3. **Aplicar Análise Multivariada Inicial**: Investigar como múltiplas variáveis interagem simultaneamente, utilizando técnicas como análise de correlação e visualizações multidimensionais.  
4. **Criar Visualizações Eficazes**: Utilizar bibliotecas como Matplotlib, Seaborn e Plotly para criar gráficos informativos e esteticamente agradáveis que comuniquem insights de forma clara.  
5. **Identificar Padrões Temporais**: Analisar como as características das turmas evoluíram ao longo dos anos e períodos disponíveis no dataset.  
6. **Segmentar e Comparar Grupos**: Comparar diferentes segmentos dos dados (ex: turmas presenciais vs. à distância, diferentes níveis de ensino, diferentes campi) para identificar diferenças significativas.

### Tarefas a Serem Realizadas no Notebook

Abra o notebook do Dia 1 e continue o trabalho criando uma nova seção para a análise exploratória.  
 Siga os passos abaixo:

#### 1\. Análise Univariada

Examine cada variável individualmente:

**Variáveis Numéricas**: Para colunas como ch\_dedicada\_periodo, capacidade\_aluno, total\_solicitacoes, ano, periodo, crie histogramas, boxplots e calcule estatísticas descritivas. Identifique a distribuição de cada variável, valores atípicos e características interessantes.

**Variáveis Categóricas**: Para colunas como nivel\_ensino, campus\_turma, tipo, distancia, crie gráficos de barras e tabelas de frequência. Analise a distribuição das categorias e identifique as mais e menos frequentes.

**Variáveis Temporais**: Para data\_inicio, data\_fim, ano, periodo, explore padrões sazonais, tendências ao longo do tempo e a duração típica das turmas.

#### 2\. Análise Bivariada

Explore relações entre pares de variáveis:

**Numérica vs. Numérica**: Crie scatter plots para examinar correlações entre variáveis como capacidade de alunos vs. carga horária, total de solicitações vs. capacidade, etc. Calcule coeficientes de correlação.

**Categórica vs. Numérica**: Utilize boxplots ou violin plots para comparar distribuições de variáveis numéricas entre diferentes categorias (ex: carga horária por nível de ensino, capacidade por campus).

**Categórica vs. Categórica**: Crie tabelas de contingência e heatmaps para explorar associações entre variáveis categóricas (ex: modalidade à distância por nível de ensino, tipo de turma por campus).

#### 3\. Análise Temporal

Investigue como as características das turmas mudaram ao longo do tempo:

Analise tendências anuais no número de turmas, capacidade média, distribuição por nível de ensino, crescimento da modalidade à distância, etc. Crie gráficos de linha temporal e identifique padrões sazonais entre os períodos letivos.

#### 4\. Análise por Segmentos

Compare diferentes grupos de turmas:

Segmente os dados por características importantes (campus, nível de ensino, modalidade) e compare métricas-chave entre os segmentos. Identifique quais campi oferecem mais turmas, quais níveis de ensino têm maior capacidade média, como a modalidade à distância se distribui, etc.

#### 5\. Identificação de Insights e Padrões

Com base nas análises anteriores, identifique e documente:

Padrões interessantes que emergiram dos dados, possíveis anomalias ou comportamentos inesperados, relações fortes entre variáveis que merecem investigação adicional, diferenças significativas entre grupos ou períodos, e tendências temporais relevantes para a gestão acadêmica.
